```python
from openai import OpenAI
from dotenv import load_dotenv
import os
```

```python
load_dotenv(override = True)
lms_api_key = os.getenv("OLLAMA_API_KEY")
lms_url = "http://localhost:1234/v1"

lms = OpenAI(base_url = lms_url, api_key="lms")
model = "llama-3.2-1b-instruct"
# reasoning_effort = "low"
```

```python
user_prompt = "Write me a 1000 words essay about love and hate."
messages = [{"role":"user","content":user_prompt}]

response = lms.chat.completions.create(model = model, messages = messages, temperature =0, seed = 0)
print(response.choices[0].message.content)
```
